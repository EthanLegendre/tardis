```python
import sklearn as sk
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import mean_absolute_error, r2_score

df = pd.read_csv("data/cleaned_dataset.csv")
column_use = ['Number of scheduled trains', 'Month', 'Departure station']
X = df[column_use]
X = pd.get_dummies(X, columns=['Departure station'], drop_first=True)
X = pd.get_dummies(X, columns=['Month'], drop_first=True)

y = df['Average delay of all trains at arrival']

X_train, X_test, y_train, y_test = sk.model_selection.train_test_split(X, y, test_size=0.2, random_state=12)
```

```python
model_rf = RandomForestRegressor(n_estimators=100, random_state=42)
model_rf.fit(X_train, y_train)
```

```python
model_logr = LogisticRegression()
model_logr.fit(X_train, y_train)
```

```python
model_lr = LinearRegression()
model_lr.fit(X_train, y_train)
```

Model evaluation

```python
y_pred = model_rf.predict(X_test)
print(y_pred)
```
